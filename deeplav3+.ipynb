```python
!mkdir -p ~/.kaggle
!cp kaggle.json ~/.kaggle/
!chmod 600 ~/.kaggle/kaggle.json
!kaggle datasets download -d fakhrealam9537/leaf-disease-segmentation-dataset

import zipfile, os, torch, torch.nn as nn, torch.optim as optim
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, Subset
from torchvision import transforms
import torchvision.transforms.functional as TF
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import random

try:
    import segmentation_models_pytorch as smp
except ImportError:
    os.system('pip install segmentation-models-pytorch')
    import segmentation_models_pytorch as smp

zip_ref = zipfile.ZipFile('/content/leaf-disease-segmentation-dataset.zip', 'r')
zip_ref.extractall('/content')
zip_ref.close()

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
DATASET_PATH = "/content/aug_data/aug_data"
SPLIT_PATH = "/content/data_split.pt"
BEST_MODEL_PATH = "/content/best_deeplab_model.pth"

BATCH_SIZE = 8
LR = 1e-4
EPOCHS = 120
IMG_SIZE = (256, 256)

class LeafDataset(Dataset):
    def __init__(self, data_dir, augment=False):
        self.img_dir = os.path.join(data_dir, "images")
        self.mask_dir = os.path.join(data_dir, "masks")
        self.files = sorted(os.listdir(self.img_dir))
        self.augment = augment
        self.norm = transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        name = self.files[idx]
        img = Image.open(os.path.join(self.img_dir, name)).convert("RGB")
        mask = Image.open(os.path.join(os.path.join(self.mask_dir, name.replace(".jpg", ".png")))).convert("L")

        img = img.resize(IMG_SIZE, Image.BILINEAR)
        mask = mask.resize(IMG_SIZE, Image.NEAREST)

        if self.augment:
            if np.random.rand() > 0.5:
                img = TF.hflip(img)
                mask = TF.hflip(mask)
            if np.random.rand() > 0.5:
                img = TF.vflip(img)
                mask = TF.vflip(mask)

        img = transforms.ToTensor()(img)
        img = self.norm(img)

        mask = transforms.ToTensor()(mask)
        mask = (mask > 0).float()

        return img, mask

torch.manual_seed(42)
np.random.seed(42)
random.seed(42)
torch.cuda.manual_seed(42)
torch.cuda.manual_seed_all(42)

torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

def mean_iou(pred, mask):
    pred = (torch.sigmoid(pred) > 0.5).float()
    inter = (pred * mask).sum((1, 2, 3))
    union = pred.sum((1, 2, 3)) + mask.sum((1, 2, 3)) - inter
    return ((inter + 1e-6) / (union + 1e-6)).mean()

full_dataset = LeafDataset(DATASET_PATH, augment=False)
total = len(full_dataset)

train_size = int(0.7 * total)
val_size = int(0.15 * total)

if os.path.exists(SPLIT_PATH):
    indices = torch.load(SPLIT_PATH)
    print("Loaded existing data split.")
else:
    indices = torch.randperm(total)
    torch.save(indices, SPLIT_PATH)
    print("Created and saved new data split.")

train_indices = indices[:train_size]
val_indices = indices[train_size:train_size + val_size]
test_indices = indices[train_size + val_size:]

train_dataset = Subset(LeafDataset(DATASET_PATH, augment=True), train_indices)
val_dataset = Subset(LeafDataset(DATASET_PATH, augment=False), val_indices)
test_dataset = Subset(LeafDataset(DATASET_PATH, augment=False), test_indices)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=2)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)

model = smp.DeepLabV3Plus(
    encoder_name="resnet50",
    encoder_weights="imagenet",
    in_channels=3,
    classes=1
).to(DEVICE)

criterion = nn.BCEWithLogitsLoss()
optimizer = optim.Adam(model.parameters(), lr=LR)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode='max',
    patience=7,
    factor=0.5
)

best_val_iou = 0.0
train_losses, val_losses, val_ious = [], [], []

for epoch in range(1, EPOCHS + 1):
    model.train()
    train_loss = 0.0

    pbar = tqdm(train_loader, desc=f"Epoch {epoch}/{EPOCHS}")
    for img, mask in pbar:
        img, mask = img.to(DEVICE), mask.to(DEVICE)

        optimizer.zero_grad()
        pred = model(img)
        loss = criterion(pred, mask)
        loss.backward()
        optimizer.step()

        train_loss += loss.item()

    train_loss /= len(train_loader)
    train_losses.append(train_loss)

    model.eval()
    val_loss = 0.0
    val_iou_epoch = 0.0
    with torch.no_grad():
        for img, mask in val_loader:
            img, mask = img.to(DEVICE), mask.to(DEVICE)
            pred = model(img)

            val_loss += criterion(pred, mask).item()
            val_iou_epoch += mean_iou(pred, mask).item()

    val_loss /= len(val_loader)
    val_iou_epoch /= len(val_loader)

    val_losses.append(val_loss)
    val_ious.append(val_iou_epoch)

    scheduler.step(val_iou_epoch)

    print(f"\nEpoch {epoch}: Train Loss = {train_loss:.4f} | Val Loss = {val_loss:.4f} | Val IoU = {val_iou_epoch:.4f}")

    if val_iou_epoch > best_val_iou:
        best_val_iou = val_iou_epoch
        torch.save(model.state_dict(), BEST_MODEL_PATH)
        print("------------>>> Best model saved <<<-------------\n")

print(f"Training Complete. Best Validation IoU: {best_val_iou:.4f}")

# ==========================================
# Graphs: Loss & IoU Visualization
# ==========================================
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ax[0].plot(train_losses, label='Train Loss')
ax[0].plot(val_losses, label='Val Loss')
ax[0].set_title('Training & Validation BCE Loss')
ax[0].set_xlabel('Epochs')
ax[0].set_ylabel('Loss')
ax[0].legend()
ax[0].grid(True)

ax[1].plot(val_ious, label='Val IoU', color='green')
ax[1].set_title('Validation IoU')
ax[1].set_xlabel('Epochs')
ax[1].set_ylabel('IoU Score')
ax[1].legend()
ax[1].grid(True)
plt.tight_layout()
plt.show()

# ==========================================
# Final Test Evaluation
# ==========================================
print("\nEvaluating best model on test set...")
model.load_state_dict(torch.load(BEST_MODEL_PATH, map_location=DEVICE))
model.eval()
test_loss = 0.0
test_iou = 0.0

with torch.no_grad():
    for img, mask in test_loader:
        img, mask = img.to(DEVICE), mask.to(DEVICE)

        pred = model(img)

        test_loss += criterion(pred, mask).item()
        test_iou += mean_iou(pred, mask).item()

test_loss /= len(test_loader)
test_iou /= len(test_loader)

print("\n\n---------------------------------")
print("---------------------------------")
print(f"\nFINAL TEST IoU:: {test_iou:.4f}")
print(f"FINAL TEST LOSS:: {test_loss:.4f}\n")
print("---------------------------------")
print("---------------------------------")

# ==========================================
# Collage of Sample Outputs (COMMENTED OUT)
# ==========================================
# print("\nGenerating collage of sample outputs...")

# Inverse normalization to display images correctly
# inv_normalize = transforms.Normalize(
#     mean=[-0.485/0.229, -0.456/0.224, -0.406/0.225],
#     std=[1/0.229, 1/0.224, 1/0.225]
# )

# model.eval()

# # Collect target_samples samples for visualization
# all_sample_images = []
# all_sample_masks = []
# target_samples = 20
# samples_collected = 0

# with torch.no_grad():
#     for img_batch, mask_batch in test_loader:
#         all_sample_images.append(img_batch)
#         all_sample_masks.append(mask_batch)
#         samples_collected += img_batch.shape[0]
#         if samples_collected >= target_samples:
#             break

#     sample_images = torch.cat(all_sample_images, dim=0)[:target_samples].to(DEVICE)
#     sample_masks = torch.cat(all_sample_masks, dim=0)[:target_samples].to(DEVICE)

#     raw_outputs = model(sample_images)
#     pred_masks = (torch.sigmoid(raw_outputs) > 0.5).float()

# num_samples_to_show = target_samples
# num_cols_per_row = 5 # Number of original-overlay pairs per row in the grid
# num_rows_for_collage = (num_samples_to_show + num_cols_per_row - 1) // num_cols_per_row # Ceiling division for rows

# # Create a figure with a grid of subplots: 2 columns for each sample (Original + Overlay)
# fig_collage, axes_collage = plt.subplots(num_rows_for_collage, num_cols_per_row * 2, figsize=(20, 4 * num_rows_for_collage))

# for i in range(num_samples_to_show):
#     row_idx = i // num_cols_per_row
#     col_start_idx = (i % num_cols_per_row) * 2

#     # Original Image
#     img_vis = inv_normalize(sample_images[i])
#     img_vis = img_vis.permute(1, 2, 0).cpu().numpy()
#     img_vis = np.clip(img_vis, 0, 1)

#     if num_rows_for_collage == 1:
#         ax_orig = axes_collage[col_start_idx]
#         ax_overlay = axes_collage[col_start_idx + 1]
#     else:
#         ax_orig = axes_collage[row_idx, col_start_idx]
#         ax_overlay = axes_collage[row_idx, col_start_idx + 1]

#     ax_orig.imshow(img_vis)
#     ax_orig.set_title(f"Sample {i+1} Original", fontsize=8)
#     ax_orig.axis("off")

#     # Segmentation Overlay
#     mask_vis = pred_masks[i].squeeze().cpu().numpy()
#     overlay = img_vis.copy()
#     red_mask = np.zeros_like(img_vis)
#     red_mask[mask_vis == 1] = [1.0, 0.0, 0.0] # Red color for the mask
#     alpha_overlay = 0.5 # Blending factor
#     overlay = img_vis * (1 - alpha_overlay) + red_mask * alpha_overlay # Blend original image with red mask
#     overlay = np.clip(overlay, 0, 1) # Ensure values are within [0, 1] after blending

#     ax_overlay.imshow(overlay)
#     ax_overlay.set_title(f"Sample {i+1} Overlay", fontsize=8)
#     ax_overlay.axis("off")

# plt.tight_layout()
# plt.suptitle("Collage of Sample Outputs (Original vs. Segmentation Overlay)", y=1.02, fontsize=16)
# plt.show()


# ==========================================
# Sample Outputs Visualization - Individual Samples
# ==========================================
print("Saving individual sample outputs:")

# Inverse normalization to display images correctly
inv_normalize = transforms.Normalize(
    mean=[-0.485/0.229, -0.456/0.224, -0.406/0.225],
    std=[1/0.229, 1/0.224, 1/0.225]
)

model.eval()

# Collect target_samples samples for visualization (if not already collected from previous collage block)
all_sample_images = []
all_sample_masks = []
target_samples = 20
samples_collected = 0

with torch.no_grad():
    for img_batch, mask_batch in test_loader:
        all_sample_images.append(img_batch)
        all_sample_masks.append(mask_batch)
        samples_collected += img_batch.shape[0]
        if samples_collected >= target_samples:
            break

    sample_images = torch.cat(all_sample_images, dim=0)[:target_samples].to(DEVICE)
    sample_masks = torch.cat(all_sample_masks, dim=0)[:target_samples].to(DEVICE)

    raw_outputs = model(sample_images)
    pred_masks = (torch.sigmoid(raw_outputs) > 0.5).float()


# Create a directory to save individual sample outputs
individual_output_dir = "/content/individual_sample_outputs"
os.makedirs(individual_output_dir, exist_ok=True)

for i in range(target_samples): # Iterate for each of the 20 samples
    fig, axes = plt.subplots(1, 4, figsize=(14, 3.5)) # Create a new figure for each sample

    # Original Image
    img_vis = inv_normalize(sample_images[i])
    img_vis = img_vis.permute(1, 2, 0).cpu().numpy()
    img_vis = np.clip(img_vis, 0, 1)

    axes[0].imshow(img_vis)
    axes[0].set_title(f"Sample {i+1} - Original Image")
    axes[0].axis("off")

    # Ground Truth Mask
    gt_mask = sample_masks[i].squeeze().cpu().numpy()

    axes[1].imshow(gt_mask, cmap="gray")
    axes[1].set_title("Ground Truth Mask")
    axes[1].axis("off")

    # Predicted Mask
    mask_vis = pred_masks[i].squeeze().cpu().numpy()

    axes[2].imshow(mask_vis, cmap="gray")
    axes[2].set_title("Predicted Mask (DeepLabV3+)")
    axes[2].axis("off")

    # Segmentation Overlay
    overlay = img_vis.copy()
    red_mask = np.zeros_like(img_vis)
    red_mask[mask_vis == 1] = [1.0, 0.0, 0.0] # Red color for the mask
    alpha = 0.5 # Blending factor
    overlay = img_vis * (1 - alpha) + red_mask * alpha # Blend original image with red mask
    overlay = np.clip(overlay, 0, 1) # Ensure values are within [0, 1] after blending

    axes[3].imshow(overlay)
    axes[3].set_title("Segmentation Overlay")
    axes[3].axis("off")

    plt.tight_layout()

    # Save each individual figure
    individual_file_name = os.path.join(individual_output_dir, f"sample_output_{i+1:02d}.png")
    plt.savefig(individual_file_name, dpi=300, bbox_inches="tight")
    plt.show() # Display the plot for this sample
    plt.close(fig) # Close the figure to free memory

print(f"Individual sample outputs saved to: {individual_output_dir}")

# ==========================================
# Save and Download All Results
# ==========================================

# Create ZIP file containing:
# 1. all_test_results
# 2. individual_sample_outputs directory

!zip -rq results_BCE.zip "{individual_output_dir}"

from google.colab import files

files.download("results_BCE.zip")

print("\n\n>>> Test images saved successfully using BEST model")
```
